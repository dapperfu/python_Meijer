```python
import requests
import time
import secrets


# Generate dynamic deviceId and timestamp
def generate_device_id(length=16):
    return secrets.token_hex(length // 2)


device_id = generate_device_id()
timestamp_ms = int(time.time() * 1000)

# Parameters and headers
params = {
    "type": "sdk-dci",
    "starttime": timestamp_ms,
    "systemVersion": "10",
    "model": "One",
    "deviceHardwareType": "qcom",
    "appIdentifier": "com.meijer.mobile.meijer",
    "deviceId": device_id,
}

headers = {
    "User-Agent": "Akamai BMPSDK/3.3.9 (Android; 10; HTC; One; en)",
    "Host": "id.meijer.com",
    "Connection": "Keep-Alive",
    "Accept-Encoding": "gzip",
    "Content-Length": "0",
}

# Making the GET request
response = requests.get(
    "https://id.meijer.com/_bm/get_params", params=params, headers=headers
)

print(response.status_code)
print(response.json())
```

```python
import requests
import secrets
import time


# Generate dynamic values
def generate_device_id(length=16):
    return secrets.token_hex(length // 2)


device_id = generate_device_id()
timestamp_ms = int(time.time() * 1000)

# Parameters
params = {
    "os": "android",
    "starttime": timestamp_ms,
    "systemVersion": "10",
    "model": "One",
    "deviceHardwareType": "qcom",
    "appIdentifier": "com.meijer.mobile.meijer",
    "deviceId": device_id,
    "serverSideSignal": decoded_str,
}

# Headers
headers = {
    "pragma": "no-cache",
    "cache-control": "no-cache",
    "user-agent": "Mozilla/5.0 (Linux; Android 10; One Build/QQ3A.200705.002; wv) AppleWebKit/537.36 (KHTML, like Gecko) Version/4.0 Chrome/81.0.4044.138 Mobile Safari/537.36",
    "accept": "*/*",
    "x-requested-with": "com.meijer.mobile.meijer",
    "sec-fetch-site": "cross-site",
    "sec-fetch-mode": "no-cors",
    "sec-fetch-dest": "script",
    "accept-encoding": "gzip, deflate",
    "accept-language": "en-US,en;q=0.9",
    "content-length": "0",
}

# Perform the request
response = requests.get(
    "https://id.meijer.com/_sec/sdk_challenge.js", params=params, headers=headers
)

print(response.status_code)
print(response.text)
```
