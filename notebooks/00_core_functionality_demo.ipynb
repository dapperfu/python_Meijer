# 🏗️ Meijer API - Core Functionality Demo

This notebook demonstrates the **complete core functionality** of the Meijer API client, showing how to use all the core classes and methods instead of notebook example functions.

## ✅ **What's Been Moved to Core**

### 🔍 **Search Functionality** (MeijerSearch class)
- **`autocomplete()`** - Real-time search suggestions
- **`search()`** - Comprehensive product search
- **`track_product_click()`** - Product interaction tracking
- **`track_search_behavior()`** - Search analytics
- **`get_product_images()`** - Product image URL generation
- **`get_department_icons()`** - Department navigation assets
- **`get_sponsored_products()`** - Sponsored product retrieval
- **`get_complex_promotions()`** - Complex promotions API

### 🛒 **Cart Functionality** (MeijerCart class)
- **`add_item_to_cart()`** - User-friendly item addition with feedback
- **`update_item_quantity_by_index()`** - Update items by position
- **`remove_item_by_index()`** - Remove items by position
- **`set_cart_store()`** - Store switching with feedback
- **`demonstrate_cart_workflow()`** - Complete workflow demo

### 🏪 **Store Functionality** (Meijer client)
- **`get_stores()`** - Enhanced store search with service filtering and comprehensive options
- **`find_stores_with_services()`** - Service-based store filtering convenience method
- **`find_nearest_store_with_service()`** - Nearest store with specific service
- **`get_store_service_summary()`** - Store service availability summary

### ⚠️ **Exception Handling** (meijer.exceptions)
- **`handle_meijer_operation()`** - Generic error handler
- **`safe_meijer_call()`** - Safe API calls with defaults
- **`retry_meijer_operation()`** - Retry with exponential backoff
- **`MeijerErrorContext`** - Context manager for error handling

### 📦 **Enhanced Models**
- **`MeijerItem`** - Enhanced with state properties (favorite, cart, list)
- **`SearchResult`** - Enhanced with filtering, sorting, and analysis methods
