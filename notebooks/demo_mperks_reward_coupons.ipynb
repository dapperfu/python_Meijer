## 📦 Setup and Imports

First, let's import the necessary modules and initialize the Meijer client.

```python
import os
import sys
from typing import Optional
from datetime import datetime

# Add the parent directory to the path to import meijer module
sys.path.insert(0, os.path.dirname(os.path.dirname(os.path.abspath(__file__))))

from meijer import Meijer
from meijer.mperks import MPerksEarnedRewards, RewardCoupon, PointBalance

print("✅ Imports completed successfully!")
```

## 🔧 Initialize Meijer Client

Now let's initialize the Meijer client and mPerks functionality.

```python
# Initialize Meijer client
meijer = Meijer()
mperks = MPerksEarnedRewards(meijer)

print("✅ Meijer client initialized")
print(f"   API Base URL: {meijer.api_base_url}")
print(f"   mPerks endpoints: {list(mperks.endpoints.keys())}")
```

## 📊 Check Current Point Balance

Let's start by checking your current mPerks point balance.

```python
print("📊 Getting current point balance...")
point_balance = mperks.get_point_balance()

if point_balance:
    print(f"✅ Current points: {point_balance.total_points:,}")
    if point_balance.expiring_points:
        print(f"   Points expiring in {point_balance.expiring_days} days: {point_balance.expiring_points:,}")
    else:
        print("   No points expiring soon")
        
    # Store for later use
    current_points = point_balance.total_points
else:
    print("❌ Failed to get point balance")
    current_points = 0
```

## 🎫 Get Available Reward Coupons

Now let's see what reward coupons are available for purchase.

```python
print("🎫 Getting available reward coupons...")
available_coupons = mperks.get_available_reward_coupons()

if available_coupons:
    print(f"✅ Found {len(available_coupons)} available coupons")
    
    # Display fuel rewards first (as they're commonly used)
    fuel_coupons = [c for c in available_coupons if c.reward_coupon_type == "Fuel"]
    if fuel_coupons:
        print(f"\n🚗 Fuel Rewards ({len(fuel_coupons)} available):")
        for i, coupon in enumerate(fuel_coupons[:5], 1):  # Show first 5
            print(f"   {i}. {coupon.name} - {coupon.point_cost:,} points")
            print(f"      {coupon.description}")
            if coupon.days_until_expiry is not None:
                if coupon.days_until_expiry > 0:
                    print(f"      Expires in {coupon.days_until_expiry} days")
                else:
                    print(f"      Expired {abs(coupon.days_until_expiry)} days ago")
            print()
    
    # Display other reward types
    other_coupons = [c for c in available_coupons if c.reward_coupon_type != "Fuel"]
    if other_coupons:
        print(f"🎁 Other Rewards ({len(other_coupons)} available):")
        for i, coupon in enumerate(other_coupons[:5], 1):  # Show first 5
            print(f"   {i}. {coupon.name} - {coupon.point_cost:,} points")
            print(f"      {coupon.description}")
            print()
else:
    print("❌ Failed to get available coupons")
```

## 🔍 Find Specific Coupon by Name

Let's search for a specific coupon, like a fuel reward.

```python
print("🔍 Finding specific coupon by name...")
fuel_coupon = mperks.find_reward_coupon_by_name("fuel")

if fuel_coupon:
    print(f"✅ Found fuel coupon: {fuel_coupon.name}")
    print(f"   Coupon ID: {fuel_coupon.coupon_id}")
    print(f"   Cost: {fuel_coupon.point_cost:,} points")
    print(f"   Description: {fuel_coupon.description}")
    print(f"   Type: {fuel_coupon.reward_coupon_type}")
    print(f"   Available: {fuel_coupon.is_available}")
    
    if fuel_coupon.display_start and fuel_coupon.display_end:
        print(f"   Display Period: {fuel_coupon.display_start.strftime('%Y-%m-%d')} to {fuel_coupon.display_end.strftime('%Y-%m-%d')}")
    
    if fuel_coupon.terms_and_conditions:
        print(f"   Terms: {fuel_coupon.terms_and_conditions[:100]}...")
    
    # Check if user has enough points
    if current_points >= fuel_coupon.point_cost:
        print(f"\n✅ Sufficient points available ({current_points:,} >= {fuel_coupon.point_cost:,})")
        can_purchase = True
    else:
        print(f"\n❌ Insufficient points ({current_points:,} < {fuel_coupon.point_cost:,})")
        can_purchase = False
else:
    print("❌ No fuel coupon found")
    can_purchase = False
```

## ⏰ Check Expiring Points

Let's see if you have any points expiring soon.

```python
print("⏰ Getting expiring points...")
expiring_points = mperks.get_points_expiring(days=30)

if expiring_points is not None:
    if expiring_points:
        print(f"✅ Found {len(expiring_points)} expiring point entries")
        for i, entry in enumerate(expiring_points[:5], 1):  # Show first 5
            print(f"   {i}. {entry}")
    else:
        print("✅ No points expiring in the next 30 days")
else:
    print("❌ Failed to get expiring points")
```

## 📋 Coupon Analysis and Statistics

Let's analyze the available coupons and provide some statistics.

```python
print("📋 Coupon Analysis:")
if 'available_coupons' in locals() and available_coupons:
    coupon_types = {}
    total_cost = 0
    
    for coupon in available_coupons:
        coupon_type = coupon.reward_coupon_type or "Unknown"
        if coupon_type not in coupon_types:
            coupon_types[coupon_type] = []
        coupon_types[coupon_type].append(coupon)
        total_cost += coupon.point_cost
    
    print(f"   Total coupons available: {len(available_coupons)}")
    print(f"   Total points needed for all: {total_cost:,}")
    print(f"   Coupon types: {', '.join(coupon_types.keys())}")
    
    # Show breakdown by type
    print("\n   📊 Breakdown by type:")
    for coupon_type, coupons in sorted(coupon_types.items()):
        type_cost = sum(c.point_cost for c in coupons)
        print(f"      {coupon_type}: {len(coupons)} coupons, {type_cost:,} points total")
    
    # Show affordability analysis
    if 'current_points' in locals() and current_points > 0:
        affordable_coupons = [c for c in available_coupons if c.point_cost <= current_points]
        print(f"\n   💰 Affordability Analysis:")
        print(f"      With {current_points:,} points, you can afford {len(affordable_coupons)} out of {len(available_coupons)} coupons")
        
        if affordable_coupons:
            cheapest = min(affordable_coupons, key=lambda x: x.point_cost)
            most_expensive = max(affordable_coupons, key=lambda x: x.point_cost)
            print(f"      Cheapest affordable: {cheapest.name} ({cheapest.point_cost:,} points)")
            print(f"      Most expensive affordable: {most_expensive.name} ({most_expensive.point_cost:,} points)")
else:
    print("   ❌ No coupon data available for analysis")
```

## 🔍 Search and Filter Coupons

Let's explore some advanced search and filtering capabilities.

```python
if 'available_coupons' in locals() and available_coupons:
    print("🔍 Advanced Coupon Search and Filtering")
    print("=" * 40)
    
    # Filter by point cost range
    if 'current_points' in locals() and current_points > 0:
        affordable_coupons = [c for c in available_coupons if c.point_cost <= current_points]
        expensive_coupons = [c for c in available_coupons if c.point_cost > current_points]
        
        print(f"💰 Point-based filtering:")
        print(f"   Affordable (≤{current_points:,} points): {len(affordable_coupons)} coupons")
        print(f"   Expensive (>{current_points:,} points): {len(expensive_coupons)} coupons")
    
    # Filter by availability
    available_now = [c for c in available_coupons if c.is_available]
    expired = [c for c in available_coupons if not c.is_available]
    
    print(f"\n⏰ Availability filtering:")
    print(f"   Available now: {len(available_now)} coupons")
    print(f"   Expired: {len(expired)} coupons")
    
    # Show some examples of each category
    if available_now:
        print(f"\n   📋 Examples of available coupons:")
        for coupon in available_now[:3]:
            print(f"      • {coupon.name} ({coupon.reward_coupon_type}) - {coupon.point_cost:,} points")
    
    if expired:
        print(f"\n   ⏰ Examples of expired coupons:")
        for coupon in expired[:3]:
            print(f"      • {coupon.name} ({coupon.reward_coupon_type}) - {coupon.point_cost:,} points")
else:
    print("🔍 Advanced search skipped - no coupon data available")
```
