## [PACKAGE] Setup and Imports

First, let's import the necessary modules and initialize the Meijer client.

```python
# Import meijer module directly

from meijer import Meijer
from meijer.mperks import MPerksEarnedRewards

print("[OK] Imports completed successfully!")
```

## [WRENCH] Initialize Meijer Client

Now let's initialize the Meijer client and mPerks functionality.

```python
# Initialize Meijer client
meijer = Meijer()
mperks = MPerksEarnedRewards(meijer)

print("[OK] Meijer client initialized")
print(f"   API Base URL: {meijer.api_base_url}")
print(f"   mPerks endpoints: {list(mperks.endpoints.keys())}")
```

## [BAR] Check Current Point Balance

Let's start by checking your current mPerks point balance.

```python
print("[BAR] Getting current point balance...")
point_balance = mperks.get_point_balance()

if point_balance:
    print(f"[OK] Current points: {point_balance.total_points:,}")
    if point_balance.expiring_points:
        print(
            f"   Points expiring in {point_balance.expiring_days} days: {point_balance.expiring_points:,}"
        )
    else:
        print("   No points expiring soon")

    # Store for later use
    current_points = point_balance.total_points
else:
    print("[X] Failed to get point balance")
    current_points = 0
```

## [TICKET] Get Available Reward Coupons

Now let's see what reward coupons are available for purchase.

```python
print("[TICKET] Getting available reward coupons...")
available_coupons = mperks.get_available_reward_coupons()

if available_coupons:
    print(f"[OK] Found {len(available_coupons)} available coupons")

    # Display fuel rewards first (as they're commonly used)
    fuel_coupons = [c for c in available_coupons if c.reward_coupon_type == "Fuel"]
    if fuel_coupons:
        print(f"\n🚗 Fuel Rewards ({len(fuel_coupons)} available):")
        for i, coupon in enumerate(fuel_coupons[:5], 1):  # Show first 5
            print(f"   {i}. {coupon.name} - {coupon.point_cost:,} points")
            print(f"      {coupon.description}")
            if coupon.days_until_expiry is not None:
                if coupon.days_until_expiry > 0:
                    print(f"      Expires in {coupon.days_until_expiry} days")
                else:
                    print(f"      Expired {abs(coupon.days_until_expiry)} days ago")
            print()

    # Display other reward types
    other_coupons = [c for c in available_coupons if c.reward_coupon_type != "Fuel"]
    if other_coupons:
        print(f"[GIFT] Other Rewards ({len(other_coupons)} available):")
        for i, coupon in enumerate(other_coupons[:5], 1):  # Show first 5
            print(f"   {i}. {coupon.name} - {coupon.point_cost:,} points")
            print(f"      {coupon.description}")
            print()
else:
    print("[X] Failed to get available coupons")
```

## [MAGNIFYING] Find Specific Coupon by Name

Let's search for a specific coupon, like a fuel reward.

```python
print("[MAGNIFYING] Finding specific coupon by name...")
fuel_coupon = mperks.find_reward_coupon_by_name("fuel")

if fuel_coupon:
    print(f"[OK] Found fuel coupon: {fuel_coupon.name}")
    print(f"   Coupon ID: {fuel_coupon.coupon_id}")
    print(f"   Cost: {fuel_coupon.point_cost:,} points")
    print(f"   Description: {fuel_coupon.description}")
    print(f"   Type: {fuel_coupon.reward_coupon_type}")
    print(f"   Available: {fuel_coupon.is_available}")

    if fuel_coupon.display_start and fuel_coupon.display_end:
        print(
            f"   Display Period: {fuel_coupon.display_start.strftime('%Y-%m-%d')} to {fuel_coupon.display_end.strftime('%Y-%m-%d')}"
        )

    if fuel_coupon.terms_and_conditions:
        print(f"   Terms: {fuel_coupon.terms_and_conditions[:100]}...")

    # Check if user has enough points
    if current_points >= fuel_coupon.point_cost:
        print(
            f"\n[OK] Sufficient points available ({current_points:,} >= {fuel_coupon.point_cost:,})"
        )
        can_purchase = True
    else:
        print(
            f"\n[X] Insufficient points ({current_points:,} < {fuel_coupon.point_cost:,})"
        )
        can_purchase = False
else:
    print("[X] No fuel coupon found")
    can_purchase = False
```

## [TIME] Check Expiring Points

Let's see if you have any points expiring soon.

```python
print("[TIME] Getting expiring points...")
expiring_points = mperks.get_points_expiring(days=30)

if expiring_points is not None:
    if expiring_points:
        print(f"[OK] Found {len(expiring_points)} expiring point entries")
        for i, entry in enumerate(expiring_points[:5], 1):  # Show first 5
            print(f"   {i}. {entry}")
    else:
        print("[OK] No points expiring in the next 30 days")
else:
    print("[X] Failed to get expiring points")
```

## [CLIPBOARD] Coupon Analysis and Statistics

Let's analyze the available coupons and provide some statistics.

```python
print("[CLIPBOARD] Coupon Analysis:")
if "available_coupons" in locals() and available_coupons:
    coupon_types = {}
    total_cost = 0

    for coupon in available_coupons:
        coupon_type = coupon.reward_coupon_type or "Unknown"
        if coupon_type not in coupon_types:
            coupon_types[coupon_type] = []
        coupon_types[coupon_type].append(coupon)
        total_cost += coupon.point_cost

    print(f"   Total coupons available: {len(available_coupons)}")
    print(f"   Total points needed for all: {total_cost:,}")
    print(f"   Coupon types: {', '.join(coupon_types.keys())}")

    # Show breakdown by type
    print("\n   [BAR] Breakdown by type:")
    for coupon_type, coupons in sorted(coupon_types.items()):
        type_cost = sum(c.point_cost for c in coupons)
        print(
            f"      {coupon_type}: {len(coupons)} coupons, {type_cost:,} points total"
        )

    # Show affordability analysis
    if "current_points" in locals() and current_points > 0:
        affordable_coupons = [
            c for c in available_coupons if c.point_cost <= current_points
        ]
        print("\n   [MONEYBAG] Affordability Analysis:")
        print(
            f"      With {current_points:,} points, you can afford {len(affordable_coupons)} out of {len(available_coupons)} coupons"
        )

        if affordable_coupons:
            cheapest = min(affordable_coupons, key=lambda x: x.point_cost)
            most_expensive = max(affordable_coupons, key=lambda x: x.point_cost)
            print(
                f"      Cheapest affordable: {cheapest.name} ({cheapest.point_cost:,} points)"
            )
            print(
                f"      Most expensive affordable: {most_expensive.name} ({most_expensive.point_cost:,} points)"
            )
else:
    print("   [X] No coupon data available for analysis")
```

## [TARGET] Specific Fuel Coupon Analysis

Let's look specifically for the fuel coupon mentioned in the user's request: "Save $1.00/gal on fuel".

```python
print("[TARGET] Specific Fuel Coupon Analysis")
print("=" * 40)

# Find the specific fuel coupon mentioned: "Save $1.00/gal on fuel"
print("[MAGNIFYING] Looking for 'Save $1.00/gal on fuel' coupon...")
specific_fuel_coupon = mperks.find_reward_coupon_by_name("Save $1.00/gal on fuel")

if specific_fuel_coupon:
    print(f"[OK] Found coupon: {specific_fuel_coupon.name}")
    print(f"   Coupon ID: {specific_fuel_coupon.coupon_id}")
    print(f"   Description: {specific_fuel_coupon.description}")
    print(f"   Point Cost: {specific_fuel_coupon.point_cost:,} points")
    print(f"   Type: {specific_fuel_coupon.reward_coupon_type}")
    print(f"   Available: {specific_fuel_coupon.is_available}")

    if specific_fuel_coupon.display_start and specific_fuel_coupon.display_end:
        print(
            f"   Display Period: {specific_fuel_coupon.display_start.strftime('%Y-%m-%d')} to {specific_fuel_coupon.display_end.strftime('%Y-%m-%d')}"
        )

    if specific_fuel_coupon.terms_and_conditions:
        print(f"   Terms: {specific_fuel_coupon.terms_and_conditions[:150]}...")

    # Check if this matches the coupon ID from the log (6521640)
    if specific_fuel_coupon.coupon_id == 6521640:
        print("\n   [TARGET] This matches the coupon ID from the log analysis!")
    else:
        print(
            f"\n   [WARN]  Coupon ID {specific_fuel_coupon.coupon_id} doesn't match expected 6521640"
        )

    # Show purchase simulation
    print(
        f"\n[CREDIT] Purchase Simulation for Coupon {specific_fuel_coupon.coupon_id}:"
    )
    print("   POST /digital/mperks40/customer/v1/rewardcoupons/available/6521640/buy")
    print("   Body: {'storeId': 71, 'cartIsActive': true}")
    print("   Headers: Authorization, Content-Type: application/json")

else:
    print("[X] Could not find the specific fuel coupon")

    # Try to find any fuel-related coupons
    print("\n[MAGNIFYING] Looking for any fuel-related coupons...")
    if "available_coupons" in locals() and available_coupons:
        fuel_coupons = [
            c
            for c in available_coupons
            if "fuel" in c.name.lower() or c.reward_coupon_type == "Fuel"
        ]

        if fuel_coupons:
            print(f"   Found {len(fuel_coupons)} fuel-related coupons:")
            for coupon in fuel_coupons:
                print(
                    f"      - {coupon.name} (ID: {coupon.coupon_id}) - {coupon.point_cost:,} points"
                )
        else:
            print("   No fuel coupons found")
```

## [MAGNIFYING] Search and Filter Coupons

Let's explore some advanced search and filtering capabilities.

```python
if "available_coupons" in locals() and available_coupons:
    print("[MAGNIFYING] Advanced Coupon Search and Filtering")
    print("=" * 40)

    # Filter by point cost range
    if "current_points" in locals() and current_points > 0:
        affordable_coupons = [
            c for c in available_coupons if c.point_cost <= current_points
        ]
        expensive_coupons = [
            c for c in available_coupons if c.point_cost > current_points
        ]

        print("[MONEYBAG] Point-based filtering:")
        print(
            f"   Affordable (≤{current_points:,} points): {len(affordable_coupons)} coupons"
        )
        print(
            f"   Expensive (>{current_points:,} points): {len(expensive_coupons)} coupons"
        )

    # Filter by availability
    available_now = [c for c in available_coupons if c.is_available]
    expired = [c for c in available_coupons if not c.is_available]

    print("\n[TIME] Availability filtering:")
    print(f"   Available now: {len(available_now)} coupons")
    print(f"   Expired: {len(expired)} coupons")

    # Show some examples of each category
    if available_now:
        print("\n   [CLIPBOARD] Examples of available coupons:")
        for coupon in available_now[:3]:
            print(
                f"      - {coupon.name} ({coupon.reward_coupon_type}) - {coupon.point_cost:,} points"
            )

    if expired:
        print("\n   [TIME] Examples of expired coupons:")
        for coupon in expired[:3]:
            print(
                f"      - {coupon.name} ({coupon.reward_coupon_type}) - {coupon.point_cost:,} points"
            )
else:
    print("[MAGNIFYING] Advanced search skipped - no coupon data available")
```

## mPerks Points History

Access your complete mPerks points transaction history including earnings, redemptions, and bonus points.

```python
# Get mPerks points history
try:
    history = meijer.mperks.history

    if history and len(history) > 0:
        print(f"📊 Found {len(history)} history events:")
        print(f"   💰 Total Earnings: {history.total_earnings:,} points")
        print(f"   🎁 Total Redemptions: {history.total_redemptions:,} points")
        print(f"   📈 Net Points: {history.net_points:,} points")
        print()

        # Show recent events
        print("📅 Recent History Events:")
        for i, event in enumerate(history[:5], 1):  # Show first 5
            print(f"  {i}. {event}")
            if event.description:
                print(f"     📝 {event.description}")
            print()

        # Filter by event type
        print("🏷️ Events by Type:")
        event_types = set(event.event_type for event in history.events)
        for event_type in sorted(event_types):
            events_of_type = history.get_events_by_type(event_type)
            total_points = sum(event.points_change for event in events_of_type)
            print(
                f"   {event_type}: {len(events_of_type)} events, {total_points:+,} points"
            )

        print()

        # Demonstrate history access methods
        print("🔍 History Access Methods:")
        print(f"   First event: {history[0]}")
        if len(history) > 1:
            print(f"   Second event: {history[1]}")
        print(f"   Total events: {len(history)}")

    else:
        print("ℹ️ No history data available or all endpoints failed")
        print("💡 This might be normal if there are no recent transactions")

except Exception as e:
    print(f"❌ Failed to get mPerks history: {e}")
```
