```python
from meijer import Meijer
from meijer.mperks import MPerksEarnedRewards

# Initialize
meijer = Meijer()
mperks = MPerksEarnedRewards(meijer)
```

```python
mperks = meijer.mperks
```

```python
# Check balance
balance = mperks.get_point_balance()
print(f"Current points: {balance.total_points:,}")
```

```python
# Browse available coupons
coupons = mperks.get_available_reward_coupons()

# Find specific fuel coupon
fuel_coupon = mperks.find_reward_coupon_by_name("$1.00/gal")
```

```python
fuel_coupon
```

```python
# Purchase (when ready)
if fuel_coupon and balance.total_points >= fuel_coupon.point_cost:
    success = mperks.buy_reward_coupon(
        coupon_id=fuel_coupon.coupon_id,
        store_id=71,
        cart_is_active=True
    )
```
